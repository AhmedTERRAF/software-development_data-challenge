# Explore the sample frames

A quick look at the raw input data before touching `astrolab/pipeline.py`.

`data/frames/` is generated, not committed (see `astrolab/synth.py`) —
running the next cell regenerates it deterministically (seed 42) if it's
not there yet, same as `python -m astrolab.pipeline` does.

```python
import sys
from pathlib import Path

# no packaging config yet (that's a backlog item) - run from the repo root on sys.path
repo_root = str(Path.cwd().parent)
if repo_root not in sys.path:
    sys.path.insert(0, repo_root)
```

```python
import matplotlib.pyplot as plt

from astrolab.io import load_frame_set
from astrolab.pipeline import _ensure_sample_frames

_ensure_sample_frames()
frames = load_frame_set()
print(f"loaded {len(frames)} frames of shape {frames[0].shape}")
```

Five independent noisy exposures of the *same* star field — same star
positions/brightnesses, different noise draw each time. That's the raw
material `stack_frames` (backlog item #1) needs to combine.

```python
fig, axes = plt.subplots(1, len(frames), figsize=(3 * len(frames), 3))
for ax, frame in zip(axes, frames):
    ax.imshow(frame, cmap="gray")
    ax.axis("off")
fig.tight_layout()
```
